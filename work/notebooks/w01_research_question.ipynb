## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: which content items a reviewer should check first for a possible refresh, before the next monthly report shows the decline has already happened. Who acts: a content/SEO reviewer with limited review time each week, who currently has no way to prioritize except scanning dashboards after the drop already shows up. The action: spend that limited time on the items the model flags first, instead of reviewing everything or reacting only after traffic has visibly fallen. Cost of a wrong call, in both directions: a false positive wastes a reviewer's time checking a page that was never actually declining — annoying but cheap, and it compounds if the queue is noisy. A false negative is worse: a real decline goes unflagged and keeps losing impressions/clicks for weeks before a dashboard finally shows it, which is exactly the loss this project exists to avoid. That asymmetry is why I care about precision/recall on the declining class specifically, not just overall accuracy — and why the group-split numbers from notebook 03 (0.815 precision / 0.790 recall on class 1, versus the optimistic 0.852/0.856 from a random split) are the honest numbers to design around, not the inflated ones.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
url = "https://raw.githubusercontent.com/Fasiuu/FLYRANK_TASK1/refs/heads/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

df.head()

decline_rate_all = (df['trend_direction'] == 'down').mean() * 100
high_vol = df['impressions_90d'] >= 500
decline_rate_high_vol = df.loc[high_vol, 'trend_direction'].eq('down').mean() * 100
decline_rate_low_vol = df.loc[~high_vol, 'trend_direction'].eq('down').mean() * 100
print(f"1) {decline_rate_all:.1f}% of all {len(df):,} starter rows are already labeled a decline (trend_direction == 'down')")
print(f"2) among the {high_vol.sum():,} higher-volume pages (impressions_90d >= 500), the decline rate is {decline_rate_high_vol:.1f}% vs {decline_rate_low_vol:.1f}% for lower-volume pages, so the decline isn't just noise in low-traffic pages")
print(f"3) this spans {df['client_id'].nunique()} distinct clients, so it's a widespread pattern, not one account's problem")
```
